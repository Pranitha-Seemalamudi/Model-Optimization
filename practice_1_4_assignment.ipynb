```python
from sklearn.datasets import load_wine

import numpy as np, pandas as pd

data = load_wine(as_frame=True)

df = data.frame.rename(columns={'target': 'wine_class'})

target_col = 'wine_class'

rng = np.random.RandomState(7)

missing_idx = rng.choice(df.index, size=25, replace=False)

missing_cols = rng.choice([c for c in df.columns if c != target_col], size=25)

for i, c in zip(missing_idx, missing_cols):

    df.loc[i, c] = np.nan

outlier_idx = rng.choice(df.index, size=5, replace=False)

df.loc[outlier_idx, 'proline'] = df['proline'].max() * rng.uniform(2.5, 4, size=5)

df.head()
```

TRAIN & TEST

```python
x=df.drop(columns=['wine_class'])
y=df['wine_class']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)

```

```python
#df.head()
df.info()
```

```python
from sklearn.linear_model import LogisticRegression
lg=LogisticRegression()
lg.fit(x_train,y_train)
lg_pred=lg.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score, recall_score,f1_score,confusion_matrix,classification_report
lg_acc=accuracy_score(y_test,lg_pred)
```

```python
lg_pr=precision_score(y_test,lg_pred,average='weighted')
lg_pr
```

```python
lg_rc=recall_score(y_test,lg_pred,average='weighted')
```

```python
lg_con=confusion_matrix(y_test,lg_pred)
lg_con
```

```python
lg_cla=classification_report(y_test,lg_pred)
lg_cla
```

RAMDOM FOREST

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier(n_estimators=100,max_depth=5)
rf.fit(x_train,y_train)
lg_pred=rf.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score, recall_score,f1_score,confusion_matrix,classification_report
rf_acc=accuracy_score(y_test,lg_pred)
rf_acc
```

```python
rf_pr=precision_score(y_test,lg_pred,average='weighted')
rf_pr
```

```python
rf_rc=recall_score(y_test,lg_pred,average='weighted')
rf_rc
```
